```python
def get_numbers(N):
    """
    N以下の正の整数のうち、3の倍数でも4の倍数でもないもののリストを返す。
    """
    # 1からNまでのループを回し、3で割り切れない かつ 4で割り切れない数だけをリストに格納する
    return [i for i in range(1, N + 1) if i % 3 != 0 and i % 4 != 0]

# --- 実行例 ---
if __name__ == "__main__":
    N = 10
    result = get_numbers(N)
    print(f"N={N} の結果: {result}") 
    # 出力結果: N=10 の結果: [1, 2, 5, 7, 10]
```
